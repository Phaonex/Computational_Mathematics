the parents is randomly combined; that is, part of the genetic material
of the father and part of the genetic material of the mother are combined

### Genetic drift

A mutation is a stable change
in the genetic material, brought about by various
means, transmitted to offspring. Mutations to
nonessential portions of the DNA are useful
for measuring time (the molecular clock).

```julia
P^Y
```

If P is the probability that a single segment of
nonessential DNA has no mutations in a year, then PY is the probability of no
mutations in a segment of DNA happening over Y years.

Assuming that mutations are so
rare that it is very unlikely that a mutation in the same segment has occurred
in the two individuals, the percentage of segments that are mutated in one or
the other is, on average

```julia
2(1- P^Y)
```

This is an estimate of the percentage of
segments that would be found different if comparing two individuals with
a common ancestor Y years ago. Using this kind of probabilistic inference,
we can estimate that the most recent common female ancestor of all living
humans lived about 150,000 years ago.

### universal HIV testing.

About 1% of the time, HIV tests give a false-positive result.
Of those who have HIV, their tests will come out positive 95% of the time.
If someone has a positive result, what is the probability that that person has
HIV?

```julia
population = 300_000_000
infected = 500_000
```

Let’s look at the numbers: Let’s say the population of the United States
is about 300,000,000, of which about 500,000 people are HIV-positive.

```julia
infected - population / population
```

Of the 500,000 who actually have the disease, the test will come out positive 95% of the time, which equals 475,000 cases.

```julia
infected - (.95 + infected)
```

```julia
tested_positive = infected * 0.95
```

There are 299,500,000 that is:

```julia
tested_false_positive = population - infected
```

```julia
tested_false_positive / population ~ .01
```

people who do not have the disease, the test will come out
falsely positive 1% of the time, which equals 2,995,000 cases. Thus, the total
number of people receiving a positive test result is:

```julia
pop_positive = tested_positive + tested_false_positive
```

Therefore, if you get a positive test result, your probability of having the disease is

```julia
tested_positive / pop_positive * 100 ~ .15 
```
