**On average the bettor will lose 0.53 cents per bet, Expected value is an average.**

We have a collection of
outcomes, and we have a probability for each outcome’s occurring. Each
outcome has a value associated with it. In this case, for 13, the value is $350,
and for the other 37 numbers, it is –$10 (the money bet on the non-winning
number).

So lets $O_1, O_2, O_3, O_4$... Be the possible outcome.

Let $P(O)$ denote the probability of an outcome and $V(O)$ denote the value if an outcome.

Then we expect value is:

```julia
sum(P.*(O) * V .* (O) + P.*(O) * V .* (O) + P.*(O) * V .* (O) + P.*(O) * V .* (O))
```

So we can calculate lost or gain from the equation above like this:

```julia
length(Ω) - 18 # if we get 18/38 we lose 20/38
```

```julia
println("$(prob(18, length(Ω)) * (10.000) + prob(20, length(Ω)) * (-10.000) ) = $(round(-0.5263157894736841, sigdigits=2))\$")
```

Casinos count on the Law of large Numbers to ensure their profits, as the table of roulette simulations illustrates:

```julia
ten_thousands_bets = map(r -> count(c -> c == 18, rand(Ω, 10_000)), 1:10);
one_milions_bets = map(r -> count(c -> c == 18, rand(Ω, 1_000_000)), 1:10);
```

```julia
ten_thousands_wins = ten_thousands_bets
ten_thousands_lost = length(ten_thousands_bets) .- ten_thousands_wins;
```

```julia
ten_thousands_avg = prob(ten_thousands_wins, 10_000) * (10.000) + prob(ten_thousands_lost , 10_000) * (-10.000);
```

```julia
one_milions_wins = one_milions_bets
one_milions_lost = length(one_milions_bets) .- one_milions_wins;
```

```julia
one_milions_avg = prob(one_milions_wins, 1_000_000) * (10.000) + prob(one_milions_lost , 1_000_000) * (-10.000);
```

```julia
roulet_bets_df = DataFrame(repetitions = 1:10, ten_thousands_avg = ten_thousands_avg, one_milions_avg = one_milions_avg)
```

```julia
println("$(round(mean(roulet_bets_df.ten_thousands_avg), sigdigits=2))\$")
```

```julia
println("$(round(mean(roulet_bets_df.one_milions_avg), sigdigits=2))\$")
```

Let us look at unexpected instances of expected value. Suppose someone
plays roulette 35 times, betting on one number each time. The expected
value of each bet is –$0.53. And the expected total value of the 35 rounds

```julia
latexify(35((1//38) * 350 + (37//38)*(-10)) ~ 35((1/38) * 350 + (37/38)*(-10)))
```

Surprisingly, the probability that a bettor would be ahead after 35 rounds is:

```julia
latexify(1 - big((37//38))^35 ~ 1 - (37/38)^35)
```

However, the bettors who are ahead are only slightly ahead, and the people
who are behind have lost $350. **Because the expected value gives weight, the
expected value is negative.**
