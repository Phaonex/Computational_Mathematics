## Cards count.

```julia
suits = ['♣', '♢', '♡', '♠']
```

```julia
face_cards = ['J', 'K', 'Q', 'A']
```

```julia
numbers = [2:10;]
```

```julia
ranks = union(numbers, face_cards)
```

```julia
colors = ["black", "red"]
```

```julia
cards_df = DataFrame(suits = suits, color=repeat(colors, 2), ranks=map(c -> map(s -> s, ranks), suits))
```

What are the changes to get all for Aces?

```julia
collect(pairs(cards_df[:, :3]))# for each values we have for possibilities
```

because we know the total number of cards to be 52 we can computer the possible hands containing all 4 aces.

```julia
cards = collect(Base.Flatten(cards_df.ranks))
```

```julia
factorial(big(length(cards))) ~ prod(1:length(cards)) # for all cards permutations.
```

```julia
size(cards_df)[1] * length(cards_df.ranks[1]) # can see above that we have 13 posibles ranks and for possibe for each numbers and faces.
```

```julia
length(cards) - 4 # could be  the fith card in five-card hand
```

So we have the calculation for the numbers of possible hands.

```julia
prod(48:length(cards))
```

```julia
hands_df = DataFrame(map(r -> (suits = cards_df[rand(1:4), 1:1][1], color = cards_df[rand(1:4), 2:2][1], ranks = cards_df[rand(1:4), :3][rand(1:13)]), [1:5;]))
```

```julia
hands = size(hands_df)[1]
```

```julia
factorial(hands) # we have also to take in concideration that some cards will be repeated so we use for non repeated 5 permutations.
```
